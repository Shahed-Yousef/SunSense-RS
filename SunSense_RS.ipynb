```python
# SunSense RS System 
# This program monitors solar panel voltage and power under Kuwait's weather conditions.
# It enables users to generate or load data, analyze readings, 
# and identify normal, warning, and critical states.
# The system helps evaluate how environmental factors 
# such as temperature and humidity impact solar panel performance and efficiency.


#import random module to generate random sensor data
import random
#Import math module for calculations
import math
#Import matplotlib to create graphs and visualize data
import matplotlib.pyplot as plt
#Import csv module to read data from CSV files
import csv
CURRENT = 10 #The current for SunSense system is fixed

#The first function
#Function to display the main menu of the SunSense RS System
#It shows all available options for the user
def display_menu():
    print("\n====================================")
    print("  Welcome to SunSense RS System")
    print("  Solar Panel Monitoring System")
    print("====================================")
    print("1. Generate Sensor Data")
    print("2. Threshold Levels")
    print("3. View Raw Data")
    print("4. Calculate Statistics")
    print("5. Threshold Monitoring")
    print("6. Compare Scenarios")
    print("7. Plot Data")
    print("8. Exit")
    
#The second function
def generate_data():
    voltage_data = [] #list to store voltage readings
    power_data = [] #list to store power readings
    
    n = int(input("Enter number of readings: "))
    #generate readings using the for loop
    for i in range(n):
        voltage = random.randint(10, 22) #generate random voltage between 10 and 22
        power = voltage * CURRENT #calculate the power using this formula
        #The values are stored in lists
        voltage_data.append(voltage)
        power_data.append(power)
    #return both lists   
    return voltage_data, power_data 
    
#Function to read voltage and power data from a CSV file    
def read_data():
    #Create empty lists to store data
    voltage_data = []
    power_data = []
    #ask user for file name
    filename = input("Enter CSV file name: ")
    
    file = open(filename, 'r')  
    reader = csv.reader(file)
    
    #loop through each row in the file
    for row in reader:
       
       if row[0].lower() == "voltage":
           continue
       #check if row has missing values    
       if len(row) < 2:
           print("Missing value. Skipping row.")
           continue
       #convert values from string to float    
       voltage = float(row[0])
       power = float(row[1])
        
       #store values in lists
       voltage_data.append(voltage)
       power_data.append(power)
        
    #close the file after reading
    file.close()
    print("Data loaded successfully!")
    
    #Return both lists
    return voltage_data, power_data

    
#The third function    
#Function to display the generated voltage and power in a clear format
def view_data(voltage_data, power_data):
    if not voltage_data:
        print("No data available. Please generate data first.")
    else:
        for i in range(len(voltage_data)):
            print(f"Reading {i+1}: Voltage = {voltage_data[i]} V | Power = {power_data[i]} W")
            
#The fourth function            
#This function analyzes voltage data and calculates minimum, maximum, average and standard deviation            
def calculate_statistics(voltage_data):
    #check if data exist
    if len(voltage_data) == 0:
        print("No data available!")
        return
    #minimum and maximum values     
    minimum = voltage_data[0] 
    maximum = voltage_data[0]
    total = 0

    for v in voltage_data:
        if v < minimum:
            minimum = v
        if v > maximum:
            maximum = v
        total += v
    average = total / len(voltage_data) #average value 
    #calculate variance 
    variance = 0
    for v in voltage_data:
        variance += (v - average) ** 2
    variance /= len(voltage_data)
    
    #standard deviation
    std_dev = math.sqrt(variance)
    #print results
    print("\n--- Statistics ---")
    print(f"Min Voltage: {minimum}")
    print(f"Max Voltage: {maximum}")
    print(f"Average Voltage: {average:.2f}")
    print(f"Standard Deviation: {std_dev:.2f}")
        
#Tuple ranges for voltage thresholds
NORMAL_RANGE = (15, 22)
WARNING_RANGE = (10, 15)
#The fifth function 
#Function to check voltage levels and classify them as Normal, Warning, or Critical
def threshold_monitoring(voltage_data):
    #Check if there is no data
    if not voltage_data:
        print("No data available. Please generate data first.")
    else:
        warning_count = 0
        critical_count = 0
        #Loop through all voltage readings
        for i in range(len(voltage_data)):
            voltage = voltage_data[i]

            #Check voltage level and assign status
            if NORMAL_RANGE[0] <= voltage <= NORMAL_RANGE[1]:
                status = "Normal"

            elif WARNING_RANGE[0] <= voltage < WARNING_RANGE[1]:
                status = "Warning"
                warning_count += 1

            else:
                status = "Critical"
                critical_count += 1

            #Print each reading with its status
            print("Reading", i + 1, ": Voltage =", voltage, "V :", status)
        #Print total counts of warnings and critical values
        print("Total Warnings:", warning_count)
        print("Total Critical:", critical_count)
        
#The sixth function
#Function to compare solar panel performance under different weather conditions
def compare_scenarios():

    normal = []
    humid = []
    fault = []

    for i in range(5):
        normal.append(random.randint(15, 22)) # Normal
        humid.append(random.randint(10, 15)) # Warning / humid

        if i % 2 == 0:
            fault.append(random.randint(0, 9))  # below 10
        else:
            fault.append(random.randint(23, 30)) # above 22

    normal_avg = sum(normal) / len(normal)
    humid_avg = sum(humid) / len(humid)
    fault_avg = sum(fault) / len(fault)

    print("\n--- Scenario Comparison ---")
    print("Normal Average:", normal_avg)
    print("Humid Average:" , humid_avg)
    print("Fault Average:" , fault_avg)         

#The 7 function
#Function to display predefined voltage threshold levels
def display_thresholds():

    print("\n--- Threshold levels ---")
    print("Normal: 15 - 22 V")
    print("Warning: 10 - 15 V")
    print("Critical: below 10 V or above 22 V")

#The 8 function
#Function to plot voltage readings using yellow and blue colors
def plot_data(voltage_data):
    #check if there is data available
    if not voltage_data:
        print("No data available. Please generate data first.")

    else:
        x = list(range(len(voltage_data))) #Create x-axis values (reading numbers)
        #plot data points as blue dots
        plt.scatter(x, voltage_data, color="blue")
        #plot line connecting the points in yellow
        plt.plot(x, voltage_data, color="gold")

        plt.title("Voltage Readings") #Add title to the graph
        plt.xlabel("Reading Number") #label x-axis
        plt.ylabel("Voltage (V)") #label y-axis
        #Display the plot
        plt.show()


#main function that controls the program flow and user interaction        
def main():   
    #Empty lists to save data
    voltage_data = []
    power_data = []
#loop to keep the program running until the user exits
while True:
       display_menu()
       choice = input("Enter your choice (1-8):")
        
       if choice == "1":
         method = input("G = Generate, L = Load CSV: ")
         if method.upper() == "G":
               voltage_data, power_data = generate_data()

         elif method.upper() == "L":
                 voltage_data, power_data = read_data()

         else:
              print("Invalid option.")

       elif choice == "2":
           display_thresholds()

       elif choice == "3":
         view_data(voltage_data, power_data)
         print("Data viewed , successfully!")  

       elif choice == "4":
         calculate_statistics(voltage_data)
         print("Statistics calculated successfully!") 

       elif choice == "5":
         threshold_monitoring(voltage_data)
           
       elif choice == "6":
           compare_scenarios()

       elif choice == "7":
           plot_data(voltage_data)

       elif choice == "8":
         print("Exiting program...")
         break 

       else:
         print("Invalid choice. Please try again.")


main()


        




        





                      
```
